🔹 Part A: Integer Indexing & Slicing
1️⃣ Integer Indexing & Negative Indexing
Input
A = np.array([100, 200, 300, 400, 500])

Expected Output
First two elements      → [100, 200]
Last three elements     → [300, 400, 500]
Middle element          → 300

Task
 Extract the above elements using integer indexing and negative indexing.



```python
import numpy as np

A=np.array([100,200,300,400,500])
first_two=A[0:2]
last_three=A[-3:]
middle_element=A[2]
print(first_two)
print(last_three)
print(middle_element)
```

🔹 Part D: Boolean Indexing vs np.where()
1️⃣3️⃣ Filtering vs Transforming

A = np.array([4, 9, 2, 7, 12, 3])

Expected Output
Boolean indexing → [9, 7, 12]
np.where         → [0, 1, 0, 1, 1, 0]

Task
Filter values > 5 using boolean indexing


Transform array using np.where()





```python
A = np.array([4, 9, 2, 7, 12, 3])
filtered = A[A > 5]

print(filtered)
```

1️⃣5️⃣ Labeling Data with np.where()
Input
scores = np.array([45, 60, 30, 85, 90])

Expected Output
Filtered scores → [60, 85, 90]
Labels          → ["Fail", "Pass", "Fail", "Pass", "Pass"]

Task
Filter using boolean indexing


Label using np.where()


```python
import numpy as np

scores = np.array([45, 60, 30, 85, 90])
filtered_scores = scores[scores >= 50]

print("Filtered scores →", filtered_scores)
labels = np.where(scores >= 50, "Pass", "Fail")

print("Labels →", labels)
```

```python
import numpy as np

scores = np.array([45, 60, 30, 85, 90])

# Boolean indexing
filtered_scores = scores[scores >= 50]

# np.where()
labels = np.where(scores >= 50, "Pass", "Fail")

print("Filtered scores →", filtered_scores)
print("Labels →", labels)
```

1️⃣6️⃣ Row-wise Broadcasting
Input
M = np.array([[1, 2, 3],
              [4, 5, 6]])
A = np.array([10, 20, 30])

Expected Output
Addition     → [[11, 22, 33],
                [14, 25, 36]]

Subtraction  → [[-9, -18, -27],
                [-6, -15, -24]]

Task
 Perform both operations using broadcasting.


```python
import numpy as np

M = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

A = np.array([10, 20, 30])

# Addition
addition = M + A

# Subtraction
subtraction = M - A

print("Addition:")
print(addition)

print("\nSubtraction:")
print(subtraction)
```
