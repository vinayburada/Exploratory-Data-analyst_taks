Q25. Load an Excel Workbook with Multiple Sheets and Read Two Different Sheets

```python
import pandas as pd

# Data for the 'Students' sheet (using the 'df' from previous cells, which contains student data)
# Assuming 'df' is the DataFrame loaded from students.xlsx which is suitable for this purpose
# If df is not defined, you might need to run the cell where it's created (O-y9gngtLUYF)

# Let's ensure 'df' exists; if not, load it from students.xlsx
try:
    df
except NameError:
    df = pd.read_excel("students.xlsx")

students_data = df.copy() # Using a copy of the existing student data

# Data for the 'Courses' sheet
courses_data = pd.DataFrame({
    'Course_ID': [1, 2, 3, 4],
    'Course_Name': ['Math', 'Science', 'History', 'Art'],
    'Credits': [3, 4, 3, 2]
})

# Create the Excel file with multiple sheets
with pd.ExcelWriter('school.xlsx') as writer:
    students_data.to_excel(writer, sheet_name='Students', index=False)
    courses_data.to_excel(writer, sheet_name='Courses', index=False)

print("Created school.xlsx with 'Students' and 'Courses' sheets.\n")

# Now, read the 'Students' sheet from the newly created 'school.xlsx'
students_df = pd.read_excel(
    "school.xlsx",
    sheet_name="Students"
)

print(students_df)
```

```python
import pandas as pd

teachers_df = pd.read_excel(
    "school.xlsx",
    sheet_name="Courses" # Changed to 'Courses' as 'Teachers' sheet does not exist
)

print(teachers_df)
```

```python
data = pd.read_excel(
    "school.xlsx",
    sheet_name=["Students", "Courses"]
)

students_df = data["Students"]
teachers_df = data["Courses"] # Changed from 'Teachers' to 'Courses'

print(students_df)
print(teachers_df)
```
